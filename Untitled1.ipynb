```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Create a simple binary image
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (150, 150), 255, -1)   # white square
cv2.circle(img, (100, 100), 25, 0, -1)              # black circle inside

# Define a 5x5 square structuring element
kernel = np.ones((5, 5), np.uint8)

# Apply morphological operations
erosion = cv2.erode(img, kernel, iterations=1)
dilation = cv2.dilate(img, kernel, iterations=1)
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

# Show results using matplotlib
titles = ['Original', 'Erosion', 'Dilation', 'Opening', 'Closing']
images = [img, erosion, dilation, opening, closing]

plt.figure(figsize=(10, 6))
for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
